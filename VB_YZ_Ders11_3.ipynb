```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
df = pd.read_csv("/content/Ders_11_telekom_musteri.csv")
```

```python

df['ayrildi_mi'].value_counts(normalize=True) # hedef değerimizin dağılımına bir bakıyoruz.
```

```python
veri_model = df.drop(columns=["musteri_id","musteri_no_son_hane"]) #öğrenme algoritmamıza etki etmeyeceğini düşündüğümüz sütunları çıkardık.
```

```python
veri_model.head()
```

```python

veri_model = pd.get_dummies(veri_model, columns=['sozlesme_tipi'],drop_first=True, dtype=int)
```

```python
veri_model.head()
```

```python
veri_model.info()
```

```python

X = veri_model.drop(columns=["ayrildi_mi"])

y = veri_model["ayrildi_mi"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)
```

```python

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


model = LogisticRegression()

model.fit(X_train_scaled, y_train)




```

```python
y_pred = model.predict(X_test_scaled)
```

```python
accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
precision_score(y_test,y_pred)
```

```python
recall_score(y_test,y_pred)
```

```python
f1_score(y_test,y_pred)
```
