```python
# Modelimize Hazırlık

X_dummy = pd.get_dummies(veri.drop(columns=["fiyat"]), columns=kategorik, drop_first=True, dtype=int)
#
y = veri["fiyat"]
#
print("Dummy Sonrası sütun sayısı:", X_dummy.shape)
```

```python
X_kat = veri.drop(columns=["fiyat"])
X_kat["marka"] = X_kat["marka"].astype("category")
X_kat["model"] = X_kat["model"].astype("category")
X_kat["il"] = X_kat["il"].astype("category")
X_kat["kozmetik_durum"] = X_kat["kozmetik_durum"].astype("category")
X_kat["satici_tipi"] = X_kat["satici_tipi"].astype("category")

print("Kategori tipi ile sütun sayısı:", X_kat.shape[1])
print()
print(X_kat.dtypes)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_dummy, y, test_size=0.25, random_state=45)

#
X_train_dummy , X_test_dummy, y_train_dummy, y_test_dummy = train_test_split(X_dummy, y, test_size=0.25, random_state=45)
#
print("Eğitim:", X_train.shape, "Test", X_test.shape)
print("Dummy Eğitim:", X_test_dummy.shape, "Test" , X_test_dummy.shape)
#
```

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, r2_score

naif = DummyRegressor(strategy="mean")
naif.fit(X_train, y_train)
naif_tahmin = naif.predict(X_test)
print("Ortalama Mutlak Hata:", mean_absolute_error(y_test, naif_tahmin))
print("R2 Score:", r2_score(y_test, naif_tahmin))
```

```python
from re import VERBOSE
from lightgbm import LGBMRegressor

lgb_ilk = LGBMRegressor(random_state=42 , verbose=-1)
lgb_ilk.fit(X_train, y_train)
tahmin_ilk = lgb_ilk.predict(X_test)

print("Eğitim MAE:", mean_absolute_error(y_train, lgb_ilk.predict(X_train)))
print("Test MAE:", mean_absolute_error(y_test, tahmin_ilk))
print("R2 Score:", r2_score(y_test, tahmin_ilk))
```

```python
lgb_dummy = LGBMRegressor(random_state=42 , verbose=-1)
lgb_dummy.fit(X_train_dummy, y_train)
tahmin_dummy = lgb_dummy.predict(X_test_dummy)
print("Eğitim MAE:", mean_absolute_error(y_train_dummy, lgb_dummy.predict(X_train_dummy)))
print("Test MAE:", mean_absolute_error(y_test_dummy, tahmin_dummy))
print("R2 Score:", r2_score(y_test_dummy, tahmin_dummy))
```

```python
#Hız Yarışı
import time
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor

baslangic = time.time()
gb = GradientBoostingRegressor(random_state=42, n_estimators=300, max_depth=5)
gb.fit(X_train_dummy, y_train)
sure_gb = time.time() - baslangic
print("Gradient Boosting Regressor:", sure_gb)

baslangic = time.time()
xgb = XGBRegressor(random_state=42, n_estimators=300, max_depth=5)
xgb.fit(X_train_dummy, y_train)
sure_xgb = time.time() - baslangic
print("XGBoost Regressor:", sure_xgb)

baslangic = time.time()
lgb_300_dummy = LGBMRegressor(random_state=42, n_estimators=300, max_depth=5)
lgb_300_dummy.fit(X_train_dummy, y_train)
sure_lgb_dummy = time.time() - baslangic
print("LGBM Regressor:", sure_lgb_dummy)

baslangic = time.time()
lgb_300 = LGBMRegressor(random_state=42, n_estimators=300, verbose=-1)
lgb_300.fit(X_train, y_train)
sure_lgb = time.time() - baslangic
print("LGBM Regressor:", sure_lgb)
```

```python
print("GB test MAE:" ,mean_absolute_error(y_test, gb.predict(X_test_dummy)))
print("XGB test MAE:", mean_absolute_error(y_test, xgb.predict(X_test_dummy)))
print("LGB Dummy test MAE:", mean_absolute_error(y_test, lgb_300_dummy.predict(X_test_dummy)))
print("LGB Kategorik test MAE:", mean_absolute_error(y_test, lgb_300.predict(X_test)))

```

Kodun Amacı ve Analizi: Gradient Boosting Algoritmalarında Hız Kıyaslaması
Bu çalışmada, aynı veri seti üzerinde popüler üç güçlü algoritma (Scikit-Learn Gradient Boosting, XGBoost ve LightGBM) aynı hiperparametre koşullarında (n_estimators=300, max_depth=5) eğitilerek model eğitim süreleri (time.time()) karşılaştırılmıştır.

```python
lgb_yaprak_7 = LGBMRegressor(random_state=42, n_estimators=300, max_depth=7)
lgb_yaprak_7.fit(X_train, y_train)

lgb_yaprak_31 = LGBMRegressor(random_state=42, n_estimators=300, max_depth=31)
lgb_yaprak_31.fit(X_train, y_train)

lgb_yaprak_255 = LGBMRegressor(random_state=42, n_estimators=300, max_depth=7)
lgb_yaprak_255.fit(X_train, y_train)

print("Yaprak sayısı 7:", mean_absolute_error(y_test, lgb_yaprak_7.predict(X_test)))
print("Yaprak sayısı 31:", mean_absolute_error(y_test, lgb_yaprak_31.predict(X_test)))
print("Yaprak sayısı 255:", mean_absolute_error(y_test, lgb_yaprak_255.predict(X_test)))
```

```python
#Bundan sonra num_leaves: 31 olarak ilerleyeceğiz. Hata değeri en düşk o.
```

```python
# # PARAMETRELERİMİZ 2 - min_child_samples - Bir yaprakta en az kaç ilan, data olacak.

lgb_cocuk_1 = LGBMRegressor(random_state=42, num_leaves=31, min_child_samples=1, verbose=-1)
lgb_cocuk_1.fit(X_train, y_train)

lgb_cocuk_100 = LGBMRegressor(random_state=42, num_leaves=31, min_child_samples=100, verbose=-1)
lgb_cocuk_100.fit(X_train, y_train)

lgb_cocuk_500 = LGBMRegressor(random_state=42, num_leaves=31, min_child_samples=500, verbose=-1)
lgb_cocuk_500.fit(X_train, y_train)

print("Çocuk Sayısı 1:", mean_absolute_error(y_test, lgb_cocuk_1.predict(X_test)))
print("Çocuk Sayısı 100:", mean_absolute_error(y_test, lgb_cocuk_100.predict(X_test)))
print("Çocuk Sayısı 500:", mean_absolute_error(y_test, lgb_cocuk_500.predict(X_test)))


```
