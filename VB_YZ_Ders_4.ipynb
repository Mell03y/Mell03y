```python
import numpy as np

# #0. Öğrenciler ve Sınav Notları Analizi

# #1. Veri oluşturalım
# 2 değişken: isim + 3 sınav notu
# 10 öğrenci

isimler = np.array([
    "Ahmet", "Ayşe", "Mehmet", "Zeynep", "Melike",
    "Elif", "Can", "Ece", "Burak", "Selin"
])

# Her öğrenci için 3 sınav notu
notlar = np.random.randint(30, 101, size=(10, 3))

print("Öğrenciler:")
print(isimler)

print("\nSınav Notları:")
print(notlar)


# #2. Veri oluşturalım - Not ortalamalarını yazdıralım

ortalamalar = notlar.mean(axis=1)

print("\nÖğrenci Ortalamaları:")

for i in range(len(isimler)):
    print(isimler[i], ":", ortalamalar[i])


# #3. Veri oluşturalım - Her sınavın ortalaması

sinav_ortalamalari = notlar.mean(axis=0)
np.round(ortalamalar)

print("\nSınav Ortalamaları:")
print("1. Sınav:", sinav_ortalamalari[0])
print("2. Sınav:", sinav_ortalamalari[1])
print("3. Sınav:", sinav_ortalamalari[2])


# #4. Veri oluşturalım - Sınıf geneli istatistik

print("\nSınıf Genel İstatistikleri")

print("Toplam not:", notlar.sum())
print("Genel ortalama:", notlar.mean())
print("En yüksek not:", notlar.max())
print("En düşük not:", notlar.min())
print("Standart sapma:", notlar.std())
print("Medyan:", np.median(notlar))


# #5. Harf notu ve geçti/geçemedi durumu

print("\nSonuçlar:")

for i in range(len(isimler)):

    ortalama = ortalamalar[i]

    if ortalama >= 90:
        harf = "AA"
    elif ortalama >= 80:
        harf = "BA"
    elif ortalama >= 70:
        harf = "BB"
    elif ortalama >= 60:
        harf = "CB"
    elif ortalama >= 50:
        harf = "CC"
    else:
        harf = "FF"

    if ortalama >= 50:
        durum = "Geçti"
    else:
        durum = "Kaldı"

    print(
        isimler[i],
        "- Ortalama:", round(ortalama, 2),
        "- Harf:", harf,
        "- Durum:", durum
    )
```
