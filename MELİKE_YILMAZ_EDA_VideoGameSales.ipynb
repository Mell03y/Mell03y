### Yayıncı Yoğunlaşması

İlk 10 yayıncının toplam küresel satışlardaki payı %70,15 olarak hesaplanmıştır. Bu sonuç, toplam satışların büyük bir bölümünün az sayıdaki büyük yayıncı tarafından gerçekleştirildiğini göstermektedir.

Bu nedenle veri setinde yayıncılar açısından yüksek bir satış yoğunlaşması olduğu söylenebilir. Ancak bu sonuç doğrudan tam anlamıyla bir tekel olduğunu göstermez.

```python
#Bölgesel Satışlar Arasındaki Korelasyon
```

```python
bolge_korelasyon = df[
    ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]
].corr()

bolge_korelasyon  #ölge satışlarının birbirleriyle ne kadar ilişkili olduğu
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    bolge_korelasyon,
    annot=True,
    cmap="RdPu" #color map
)

plt.title("Bölgesel Satışlar Arasındaki Korelasyon")
```

```python
##Değerler -1 ile +1 arasındadır: (korelasyon nasıl okuyacağız?)

#1'e yakın → iki bölgenin satışları birlikte artıp azalıyor.
#0'a yakın → aralarında zayıf ilişki var.
#-1'e yakın → biri artarken diğeri azalma eğiliminde.
```

```python
#Grafikte en önemli değerler:

#Kuzey Amerika – Avrupa: 0.77 → güçlü pozitif ilişki
#Avrupa – Diğer Bölgeler: 0.73 → güçlü pozitif ilişki
#Kuzey Amerika – Diğer Bölgeler: 0.63 → orta/güçlü pozitif ilişki
#Kuzey Amerika – Japonya: 0.45 → orta düzey ilişki
#Avrupa – Japonya: 0.44 → orta düzey ilişki
#Japonya – Diğer Bölgeler: 0.29 → zayıf ilişki
```

### Bölgesel Satışlar Arasındaki Korelasyon
