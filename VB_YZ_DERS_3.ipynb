```python
vize = int(input("Vize notunuzu giriniz: "))
final = int(input("Final notunuzu giriniz: "))

ortalama = vize * 0.40 + final * 0.60

print("Ortalamanız:", ortalama)

if ortalama >= 50:
    print("Geçtiniz")
else:
    print("Kaldınız")
```

```python
#Soru 3: Basit tahmin oyunu:
# Sayı gizli değeri sabit.
# Tahmin hakkı 3 tane olsun.
#inputtan sayıyı girsin.  Tahmin etsin .
# Eğer bildiyse bildi.
# Eğer sayı küçükse daha büyük sayı ver.
# Eğer sayı büyükse daha küçük sayı ver.
# Eğer hakkı biterse hakkın bitti sayı ... du.
```

```python
gizli_sayi = 19
hak = 3

while hak > 0:
    sayi = int(input("Tahmininizi giriniz: "))

    if sayi == gizli_sayi:
        print("Bildiniz!")
        break

    elif sayi < gizli_sayi:
        print("Daha büyük bir sayı gir.")

    else:
        print("Daha küçük bir sayı gir.")

    hak = hak - 1

if hak == 0:
    print("Hakkınız bitti. Sayı:", gizli_sayi)
```
